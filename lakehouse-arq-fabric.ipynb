# FASE 3: TRANSFORMAÇÃO (ETL)

```python
# 3.1 Transformação com Spark SQL 

# 1. LIMPEZA: Remove nulos, duplicatas, valida datas
spark.sql("""
CREATE OR REPLACE TABLE silver_orders AS
SELECT
    CAST(order_id AS STRING) AS order_id,
    CAST(customer_id AS STRING) AS customer_id,
    CAST(
        TO_TIMESTAMP(
            order_purchase_timestamp,
            'yyyy-MM-dd HH:mm:ss'
        ) AS DATE
    ) AS order_purchase_date,
    order_status,
    CURRENT_TIMESTAMP() AS ingestion_time
FROM raw_orders
WHERE order_id IS NOT NULL
  AND customer_id IS NOT NULL
  AND order_purchase_timestamp IS NOT NULL
  AND order_status IN ('delivered','shipped','pending','cancelled')
""")
```

```python
spark.sql("""
CREATE OR REPLACE TABLE silver_order_values AS

SELECT
    order_id,
    ROUND(SUM(price + freight_value),2) AS total_value
FROM raw_order_items
GROUP BY order_id
""")
```

```python
# 2. ENRIQUECIMENTO: Join com clientes

spark.sql("""
CREATE OR REPLACE TABLE silver_orders_enriched AS

SELECT
    o.order_id,
    o.customer_id,
    o.order_purchase_date,
    o.order_status,
    c.customer_state,
    c.customer_city,
    v.total_value,
    YEAR(o.order_purchase_date) AS year_order_date,
    MONTH(o.order_purchase_date) AS month_order,
    QUARTER(o.order_purchase_date) AS quarter_order

FROM silver_orders o

LEFT JOIN raw_customers c
ON o.customer_id = c.customer_id

LEFT JOIN silver_order_values v
ON o.order_id = v.order_id

WHERE o.order_status <> 'cancelled'
""") 
```

```python
# 3. AGREGAÇÃO: Cria métricas por cliente

spark.sql("""
CREATE OR REPLACE TABLE silver_customer_metrics AS

SELECT
    customer_id,
    customer_state,
    COUNT(order_id) AS total_orders,
    ROUND(SUM(total_value),2) AS lifetime_value,
    MAX(order_purchase_date) AS last_purchase_date,
    ROUND(AVG(total_value),2) AS avg_order_value,
    MIN(order_purchase_date) AS first_purchase_date

FROM silver_orders_enriched

GROUP BY
    customer_id,
    customer_state

HAVING COUNT(order_id) > 0

""") 
```

* 3.2 Transformação com PySpark (Dados Maiores)

```python
from pyspark.sql import functions as F

orders_raw = spark.table("raw_orders")

orders_clean = (
    orders_raw
    .filter(F.col("order_id").isNotNull())
    .filter(F.col("customer_id").isNotNull())
    .dropDuplicates(["order_id"])
    .withColumn(
        "order_purchase_date",
        F.to_date("order_purchase_timestamp")
    )
)

orders_clean.show(5)

# 2. adicionar dimensões de tempo

orders_temporal = (
    orders_clean
    .withColumn("year", F.year("order_purchase_date"))
    .withColumn("month", F.month("order_purchase_date"))
    .withColumn("quarter", F.quarter("order_purchase_date"))
    .withColumn("week", F.weekofyear("order_purchase_date"))
    .withColumn("day_of_week", F.dayofweek("order_purchase_date"))

)

# 3. Salvar como tabela
orders_temporal.write.mode('overwrite').option('mode', 'FAILFAST').saveAsTable('silver_orders_v2')
```

# FASE 4: MODELAGEM (GOLD LAYER - STAR SCHEMA)

```python

# DIM_DATE - Tabela de calendário (essencial para BI)

spark.sql("""
CREATE OR REPLACE TABLE dim_date AS

WITH dates AS (

    SELECT
        EXPLODE(
            SEQUENCE(
                TO_DATE('2016-01-01'),
                TO_DATE('2019-12-31'),
                INTERVAL 1 DAY
            )
        ) AS date_key

)

SELECT

    CAST(DATE_FORMAT(date_key,'yyyyMMdd') AS INT) AS date_id,

    date_key,

    YEAR(date_key) AS year,

    QUARTER(date_key) AS quarter,

    MONTH(date_key) AS month,

    DAY(date_key) AS day,

    WEEKOFYEAR(date_key) AS week_number,

    DAYOFWEEK(date_key) AS day_of_week,

    DATE_FORMAT(date_key,'EEEE') AS day_name,

    DATE_FORMAT(date_key,'MMMM') AS month_name,

    CASE
        WHEN MONTH(date_key) IN (1,2,3) THEN 'Q1'
        WHEN MONTH(date_key) IN (4,5,6) THEN 'Q2'
        WHEN MONTH(date_key) IN (7,8,9) THEN 'Q3'
        ELSE 'Q4'
    END AS quarter_name,

    CASE
        WHEN DAYOFWEEK(date_key) IN (1,7)
        THEN TRUE
        ELSE FALSE
    END AS is_weekend

FROM dates

ORDER BY date_key
""") 
```

```python
spark.sql(
'''
CREATE OR REPLACE TABLE im_customer AS

SELECT

     ROW_NUMBER() OVER (
        ORDER BY customer_state,
        customer_city,
        customer_id
      
      
    ) AS customer_key,

    customer_id,
    customer_city,
    customer_state,

    CONCAT(customer_city, '-', customer_state) AS customer_location,

    CURRENT_TIMESTAMP() AS creted_at

FROM raw_customers

WHERE customer_id IS NOT NULL

'''
) 
```

```python
# DIM_PRODUCT - Dimensão de produtos



spark.sql("""
CREATE OR REPLACE TABLE dim_product AS

SELECT

    ROW_NUMBER() OVER (
        ORDER BY product_category_name, product_id
    ) AS product_key,

    product_id,

    product_category_name,

    product_name_lenght,

    product_description_lenght,

    product_photos_qty,

    product_weight_g,

    product_length_cm,

    product_height_cm,

    product_width_cm,

    CASE
        WHEN product_weight_g < 500 THEN 'hight'
        WHEN product_weight_g < 2000 THEN 'Medium'
        ELSE 'wheight'
    END AS weight_category,

    CURRENT_TIMESTAMP() AS created_at

FROM raw_products

WHERE product_id IS NOT NULL
""")
```

* 4.2 Criar Fact Table

```python
# FACT_SALES - Tabela de fatos (transações)

spark.sql("""
CREATE OR REPLACE TABLE fact_sales AS

SELECT

    CAST(
        DATE_FORMAT(o.order_purchase_date,'yyyyMMdd')
        AS INT
    ) AS date_id,

    dc.customer_key,

    dp.product_key,

    o.order_id,

    oi.order_item_id,

    CAST(oi.price AS DECIMAL(10,2)) AS unit_price,

    CAST(oi.freight_value AS DECIMAL(10,2)) AS freight_value,

    CAST(
        oi.price + oi.freight_value
        AS DECIMAL(10,2)
    ) AS line_amount,

    o.order_status,

    CASE
        WHEN o.order_status = 'delivered'
        THEN 1
        ELSE 0
    END AS is_completed,

    CURRENT_TIMESTAMP() AS loaded_timestamp

FROM silver_orders_enriched o

LEFT JOIN raw_order_items oi
       ON o.order_id = oi.order_id

LEFT JOIN dim_customer dc
       ON o.customer_id = dc.customer_id

LEFT JOIN dim_product dp
       ON oi.product_id = dp.product_id

WHERE o.order_purchase_date IS NOT NULL
""")


```
